```python
#changing values of a specific location
df.loc[7,'Duration']=45
print(df)
```

```python
#replacing vaules based on a condition
for x in df.index:
    # if in the location (x: depends on loop)
    # and duration is>120
    #in the location x 
    #set value as 120 
    if df.loc[x,"Duration"]>120:
        df.loc[x,"Duration"]=120

                
```

```python
#Delete the values based on a condition
for x in df.index:
    if df.loc[x, "Duration"]>120:
        df.drop(x,inplace=True)
    
```

```python
#Drop the duplicate row 
nv=df.duplicated()
new_out=df.loc[nv==True]
print(nv)
print(new_out)
df.drop_duplicates(inplace=True)
print(df.to_string())
```
