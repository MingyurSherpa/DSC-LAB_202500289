# File handling

```python
#open(filename,mode)
# mode : --->
    #r: eopen an existing file for read op
    #w: Write op
    #a: Append (make some changes)
    #r+: Read and write(Overwrite the existing file)
    #w+: Write and read(Override the ex.file)
    #a+: to read and append the data(NOT OERRIDE)
    #x: to create a specific file
    #'t': text file
    #'b': binary file
```

```python
#to open and read a file
f=open("202500289.txt","r")
```

```python
print(f.readline())
```

```python
#to read a file line by line
f=open("202500289.txt","r")
for x in f:
         print(x)
f.close()
```

```python
#to create a file 
file=open("a2.txt","w")
file.write("this is the first lab of file handling\n")
file.write("this is the second line")
file.close()
```

```python
file=open("a2.txt","a")
file.write("this is a new line\n this is the second line")
file.close()
```

```python
#to delete a file
import os
if os.path.exists("a2.txt"):
            os.remove("a2.txt") 
            print("file deleted")
else:
       print("file not deleted")
```

```python
#python program to illustrate split()fucntion
pathh = "202500289.txt"
with open(pathh,"r") as file:
      data=file.readlines()
      for line in data:
        word=line.split()
        print(word)
```

```python
import pandas as pd
mydataset={
    'cars':["kia","honda", "maruti","jeep"], 
    'pass':[5,3,7,4]
}
myvar=pd.DataFrame(mydataset)
print(myvar)
```

```python
#to create a series
a=[1,7,2]
myvar=pd.Series(a,index=["sl.no","marks","AM"])
print(myvar)
```

```python
#create a series form a dictionary( key--> value pair )
calories={"day1":420,"day2":380,"day3":500}
myvar=pd.Series(calories)
print(myvar)
```

```python
#create a series using daa from day1 and day 2
calories={"day1":420,"day2":380,"day3":500}
myvar=pd.Series(calories, index=["day1","day2"])
print(myvar)
```

```python
data={
    "calories":[420, 380, 390],
    "days":[1,2,3],
    "duration":[50,60,70]}
df=pd.DataFrame(data)
print(df)
```
