```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
url='https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'

df=pd.read_csv(url)
print(df)
attributes=["sepal_length:","sepal_width","petal_length","petal_width","class"]
dataset= pd.read_csv(url,names=attributes)
dataset.column=attributes
dataset
```

```python
# plot histogram for Univariate 
iris['petal_length'].hist(bins=20)
# histogram pot using matplotib
plt.hist(data=dataset,x='petal_length',bins=20)
# using seaborn 
sns.histplot(dataset['petal_length'],bins=20)
#histogram (ud) f each variabes using pandas
iris.hist(bins=20,figsize=(10,10))
```

```python
# modify 
iris.groupby('species').petal_length.plot.hist(alpha=0.4)
plt.xlabel('petal length(cm)')
plt.suptitle("Histogram of petal length (cm) for different species")
plt.legend()
plt.grid()
```

```python
# box plot
sns.boxplot(data=iris,y='sepal_length')
min(iris['sepal_length'])
np.median(iris['sepal_length'])
iris.sepal_length.quantile([0.25,0.5,0.75])
np.percentile(iris.sepal_length,[25,50,70])
#boxplot for each categories
#use seaborn ,box plot and take data=iris-----> x=species----- y=sepal length
sns.boxplot(data=iris,x='species',y='sepal_length')
```
