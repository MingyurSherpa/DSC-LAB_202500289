### Binomial Data Distribution

```python
x = random.binomial(n=10,p=0.5,size=1000)
print(x)
```

```python
#Random,matplotlib,sebaorn have been imported in the prev cells
sns.displot(random.binomial(n=10,p=0.5,size=1000),kde =True)
#kde=True displays both histogram and bell curve graphs
```

### Uniform distribution

```python
from numpy import random
x = random.uniform(size = (2,3))
print(x)
```

```python
import seaborn as sns
sns.displot(random.uniform(size = (2,3)))
```

#### Poisson Distribution

```python
x = random.poisson(lam=2,size=7)
print(x)
```

```python
sns.displot(random.poisson(lam=2,size=7),kde = True)
```

#### Multinomial Distribution

```python
x = random.multinomial(n = 20,pvals = [1/6,1/6,1/6,1/6,1/6,1/6])
x = random.multinomial(20,[1/6]*6,size = 1)
print(x)
```

```python
sns.displot(random.multinomial(n = 20,pvals = [1/6,1/6,1/6,1/6,1/6,1/6]),kde = True)
sns.displot(random.multinomial(20,[1/6]*6,size=1),kde = True)
```

##### Exponential Distribution

```python
x = random.exponential(scale=100,size = (1,3))
print(x)
```

```python
sns.displot(random.exponential(size = 100))
```

###### Rayleigh Distribution

```python
x = random.rayleigh(scale = 10,size = 100)
print(x)
```

```python
sns.displot(random.rayleigh(scale = 10,size = 100),kde = True)
```
